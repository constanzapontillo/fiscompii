```python
import numpy as np #importamos numpy 
```

```python
rng = np.random.default_rng(seed = 33) #definimos una semilla
```

```python
L = np.array([1.0, 2.5, 0.8, 5.0, 10.0]) #convertimos las listas en arreglos usando array
d = np.array([10, 25, 15, 40, 60])
```

```python
def flujo_observado(L,d):  #definimos la funcion 
    lum = L * 3.828 * 10**26 #cambiamos de sistema de unidades
    mts = d * 3.086 * 10**16 #cambiamos de sistema de unidades
    flujo = lum / (4 * np.pi * mts**2) #calculas el flujo observado 
    return flujo #retorna los flujos 
```

```python
flujo = flujo_observado(L,d) #llamamos a la funcion
flujo
```

```python
idx_mayor = np.argmax(flujo) #con argmax nos dan el indice mayor de donde esta el mayor flujo
idx_menor = np.argmin(flujo) #con argmin nos dan el indice menor de donde esta el menor flujo
print(idx_mayor)
print(idx_menor)
```

```python
prom = np.mean(flujo) #la funcion de numpy np.mean nos da el promedio
mediana = np.median(flujo) #de misma manera np.median nos da la mediana
print(prom)
print(mediana)
```

```python
limite = 10**-11 
mascara = flujo >= limite #delimitamos nuestra mascara
print(mascara)
idx_limite = np.where(mascara > limite) #muestra los indices 
print(idx_limite)
```

```python
orden = np.argsort(flujo)[::-1] #ordena las estrellas de mayor a menor por eso el [::-1]
print(orden)
```

```python
t = np.linspace(0,10,500) #arreglo temporal de 10 años compuesto de 500 partes
```

```python
flujo_mascara = np.ones(500) #creamos una mascara 
ruido_gaussiano = rng.normal(loc=0,scale=0.002,size=(500)) #definimos promedio, desviacion, cant
flujo_simulacion = flujo_mascara + ruido_gaussiano
```

```python
mascara_tiempo = (t >= 4.5) & (t <= 5) #intervalo de tiempo que nos piden 
flujo_simulacion[mascara_tiempo]-= 0.02 #disminuyendo el flujo 
```

```python
promedio_curva = np.mean(flujo_simulacion) # promedio del flujo
desviacion = np.std(flujo_simulacion) # desviacion estandar
print(promedio_curva)
print(desviacion)
```

```python
minimo_valor = np.min(flujo_simulacion) # busca el valor minimo
idx_minimo = np.argmin(flujo_simulacion) # nos na el indice de este minimo valor
print(minimo_valor)
print(idx_minimo)
```

```python
tiempo_minimo = t[idx_minimo] # buscamos en el tiempo con el mismo indice que el valor minimo
print(tiempo_minimo)
```

```python
filtro = promedio_curva - 3*desviacion # esta sera una condicion
cumplen_condicion = flujo_simulacion[flujo_simulacion < filtro] # aplicamos la condicion   revisar
print(cumplen_condicion)
```

```python
cant_puntos = len(cumplen_condicion) # funcion len nos da una cantidad
print(cant_puntos)
```

```python
prom_transito = np.mean(cumplen_condicion) #promedio de punto que cumplen condicion
profundidad = 1 - prom_transito # aplicamos formula
print(profundidad)
```

```python
razon = np.sqrt(profundidad)
print(razon)
```

```python
tiempos_condicion = t[flujo_simulacion < filtro] #tiempos que usamos
deteccion_inicial = np.min(tiempos_condicion) #tiempos donde se tomaron datos
deteccion_final = np.max(tiempos_condicion)
print(f"Se detecto en en el intervalo {deteccion_inicial} - {deteccion_final}")
print("Intervalo original 4.5 - 5.0")
```
